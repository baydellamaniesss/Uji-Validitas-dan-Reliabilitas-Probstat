# Uji Validitas dan Reliabilitas Instrumen


## Langkah 1: Impor pustaka

```
import re
import pandas as pd
import numpy as np
from scipy import stats
from google.colab import files
```

## Langkah 2: Unggah berkas data
Pilih file `Blank_Quiz__Responses_.xlsx`.

```
up = files.upload()
nama_file = list(up.keys())[0]
print("File terbaca:", nama_file)
```

## Langkah 3: Membaca data dan memisahkan butir


```
JUMLAH_KOLOM_DEMOGRAFI = 6   # sesuai data ini

df = pd.read_excel(nama_file)
butir = df.iloc[:, JUMLAH_KOLOM_DEMOGRAFI:]

# Buang kolom non-Likert (mis. 'Durasi (jam) ...') yang tidak memiliki tanda " ["
dibuang = [k for k in butir.columns if " [" not in k]
butir = butir[[k for k in butir.columns if " [" in k]]
print("Kolom non-butir yang dibuang:", dibuang)

print("Jumlah responden :", len(df))
print("Jumlah butir     :", butir.shape[1])
print("Data kosong      :", int(butir.isna().sum().sum()))
```

## Langkah 4: Mengubah jawaban Likert menjadi angka

```
skala = {
    "Sangat Tidak Setuju (STS)": 1,
    "Tidak Setuju (TS)": 2,
    "Netral (N)": 3,
    "Setuju (S)": 4,
    "Sangat Setuju (SS)": 5,
}
num = butir.replace(skala).apply(pd.to_numeric, errors="coerce")

# Harus 0. Jika lebih dari 0, ada jawaban yang tidak terbaca.
print("Sel yang gagal dikonversi:", int(num.isna().sum().sum()))
num.head()
```

## Langkah 5: Mengelompokkan butir per dimensi

```
dimensi = {}
for kolom in butir.columns:
    nama = re.match(r"^(.*?) \[", kolom).group(1).strip()
    dimensi.setdefault(nama, []).append(kolom)

for nama, daftar in dimensi.items():
    print(f"{nama}: {len(daftar)} butir")
```

## Langkah 6: Uji validitas (korelasi Pearson butir-total)

```
n = len(num)
t = stats.t.ppf(0.975, n - 2)
r_tabel = t / np.sqrt((n - 2) + t**2)
print(f"n = {n} | r tabel (α=0,05; df={n-2}) = {r_tabel:.3f}")

hasil_validitas = []
for nama, daftar in dimensi.items():
    x = num[daftar]
    total = x.sum(axis=1)
    for i, kolom in enumerate(daftar, 1):
        r, p = stats.pearsonr(x[kolom], total)
        r_koreksi, _ = stats.pearsonr(x[kolom], total - x[kolom])
        hasil_validitas.append({
            "Dimensi": nama,
            "Butir": f"B{i}",
            "r hitung": round(r, 3),
            "Corrected r": round(r_koreksi, 3),
            "p-value": round(p, 4),
            "Keterangan": "Valid" if r > r_tabel else "Tidak valid",
        })

tabel_validitas = pd.DataFrame(hasil_validitas)
tabel_validitas
```

## Langkah 7: Uji reliabilitas (Cronbach's alpha)

```
def cronbach_alpha(x):
    k = x.shape[1]
    return k / (k - 1) * (1 - x.var(ddof=1).sum() / x.sum(axis=1).var(ddof=1))

def kategori(a):
    return ("Sangat baik" if a >= 0.90 else "Baik" if a >= 0.80
            else "Reliabel" if a >= 0.70 else "Tidak reliabel")

hasil_reliabilitas = []
alpha_dihapus = []

for nama, daftar in dimensi.items():
    x = num[daftar]
    a = cronbach_alpha(x)
    hasil_reliabilitas.append({
        "Dimensi": nama,
        "Jumlah butir": len(daftar),
        "Cronbach's alpha": round(a, 3),
        "Kategori": kategori(a),
    })
    for i, kolom in enumerate(daftar, 1):
        if len(daftar) > 2:
            alpha_dihapus.append({
                "Dimensi": nama,
                "Butir": f"B{i}",
                "Alpha jika dihapus": round(cronbach_alpha(x.drop(columns=kolom)), 3),
            })

tabel_reliabilitas = pd.DataFrame(hasil_reliabilitas)
tabel_alpha_dihapus = pd.DataFrame(alpha_dihapus)
display(tabel_reliabilitas)
display(tabel_alpha_dihapus)
```

```
daftar_butir = []
for nama, daftar in dimensi.items():
    for i, kolom in enumerate(daftar, 1):
        daftar_butir.append({
            "Dimensi": nama,
            "Butir": f"B{i}",
            "Pernyataan": re.search(r"\[(.*)\]", kolom).group(1),
        })
tabel_butir = pd.DataFrame(daftar_butir)
pd.set_option("display.max_colwidth", None)
tabel_butir
```

## Tambahan: Ringkasan keputusan butir (sesuai pedoman modul bagian 6)

```
alpha_dim = tabel_reliabilitas.set_index("Dimensi")["Cronbach's alpha"]
ringkas = tabel_validitas.merge(tabel_alpha_dihapus, on=["Dimensi", "Butir"], how="left")
ringkas["Alpha dimensi"] = ringkas["Dimensi"].map(alpha_dim)

def keputusan(r):
    valid = r["Keterangan"] == "Valid"
    naik = r["Alpha jika dihapus"] > r["Alpha dimensi"]
    if valid and not naik: return "Pertahankan"
    if (not valid) and naik: return "Buang atau revisi, lalu uji ulang"
    if valid and naik: return "Tinjau isi butir"
    return "Tinjau ulang sebelum dibuang"

ringkas["Keputusan"] = ringkas.apply(keputusan, axis=1)
print("Jumlah butir valid (r hitung)      :", (ringkas["Keterangan"] == "Valid").sum())
print("Jumlah butir valid (corrected r≥0,30):", (ringkas["Corrected r"] >= 0.30).sum())
ringkas
```

## Langkah 8: Menggabungkan dan mengunduh hasil

```
gabung = tabel_validitas.merge(tabel_alpha_dihapus, on=["Dimensi", "Butir"], how="left")

with pd.ExcelWriter("hasil_uji_validitas_reliabilitas.xlsx") as w:
    gabung.to_excel(w, sheet_name="Validitas", index=False)
    tabel_reliabilitas.to_excel(w, sheet_name="Reliabilitas", index=False)
    tabel_butir.to_excel(w, sheet_name="Daftar Butir", index=False)
    ringkas.to_excel(w, sheet_name="Keputusan", index=False)

files.download("hasil_uji_validitas_reliabilitas.xlsx")
```

Latihan 3: Buang satu butir, uji ulang, bandingkan alpha

```
# ===== LATIHAN 3: Buang satu butir, uji ulang, bandingkan alpha =====
dim_target = "Penggunaan AI pada tugas atau projek kuliah"
idx_buang = 4                      # B4 (urutan ke-4 pada dimensi tsb)
kolom_buang = dimensi[dim_target][idx_buang - 1]
print("Butir yang dibuang:", re.search(r"\[(.*)\]", kolom_buang).group(1))

# Kamus dimensi baru tanpa butir tsb
dimensi_baru = {k: list(v) for k, v in dimensi.items()}
dimensi_baru[dim_target].remove(kolom_buang)

# --- Validitas ulang (hanya dimensi yang berubah) ---
daftar = dimensi_baru[dim_target]
x = num[daftar]
total = x.sum(axis=1)
hasil = []
for i, kolom in enumerate(daftar, 1):
    r, p = stats.pearsonr(x[kolom], total)
    rk, _ = stats.pearsonr(x[kolom], total - x[kolom])
    hasil.append({"Butir (baru)": f"B{i}", "r hitung": round(r, 3),
                  "Corrected r": round(rk, 3),
                  "Keterangan": "Valid" if r > r_tabel else "Tidak valid"})
print("\nValiditas ulang:", dim_target)
display(pd.DataFrame(hasil))

# --- Reliabilitas sebelum vs sesudah ---
a_sebelum = cronbach_alpha(num[dimensi[dim_target]])
a_sesudah = cronbach_alpha(num[dimensi_baru[dim_target]])
perbandingan = pd.DataFrame([{
    "Dimensi": dim_target,
    "Jumlah butir sebelum": len(dimensi[dim_target]),
    "Alpha sebelum": round(a_sebelum, 3),
    "Kategori sebelum": kategori(a_sebelum),
    "Jumlah butir sesudah": len(dimensi_baru[dim_target]),
    "Alpha sesudah": round(a_sesudah, 3),
    "Kategori sesudah": kategori(a_sesudah),
    "Selisih alpha": round(a_sesudah - a_sebelum, 3),
}])
display(perbandingan)

# --- Alpha jika dihapus pada dimensi baru ---
x = num[dimensi_baru[dim_target]]
display(pd.DataFrame([{"Butir (baru)": f"B{i}",
    "Alpha jika dihapus": round(cronbach_alpha(x.drop(columns=k)), 3)}
    for i, k in enumerate(x.columns, 1)]))
```

Latihan 4: Pengaruh ukuran sampel

```
# ===== LATIHAN 4: Pengaruh ukuran sampel =====
def uji_validitas(data, dimensi):
    m = len(data)
    t = stats.t.ppf(0.975, m - 2)
    rt = t / np.sqrt((m - 2) + t**2)
    valid, valid_k = 0, 0
    for daftar in dimensi.values():
        x = data[daftar]
        total = x.sum(axis=1)
        for kolom in daftar:
            if x[kolom].std() == 0:          # varians nol -> tidak bisa diuji
                continue
            r, _ = stats.pearsonr(x[kolom], total)
            rk, _ = stats.pearsonr(x[kolom], total - x[kolom])
            valid += r > rt
            valid_k += rk >= 0.30
    return rt, int(valid), int(valid_k)

total_butir = sum(len(v) for v in dimensi.values())

# (a) Contoh sesuai modul: 20 responden, random_state=1
sub = num.sample(20, random_state=1)
rt, v, vk = uji_validitas(sub, dimensi)
print(f"n=20 (random_state=1): r tabel = {rt:.3f} | valid (r hitung) = {v}/{total_butir} | valid (corrected r>=0,30) = {vk}/{total_butir}")

# (b) Variasi ukuran sampel, diulang 20 kali per ukuran lalu dirata-rata
ukuran = [10, 15, 20, 25, 30, 37]
baris = []
for m in ukuran:
    hasil = []
    for seed in range(20):
        sub = num if m == len(num) else num.sample(m, random_state=seed)
        hasil.append(uji_validitas(sub, dimensi))
    rt = hasil[0][0]
    baris.append({
        "n": m,
        "df": m - 2,
        "r tabel": round(rt, 3),
        "Rata-rata butir valid (r hitung)": round(np.mean([h[1] for h in hasil]), 1),
        "Rata-rata butir valid (corrected r>=0,30)": round(np.mean([h[2] for h in hasil]), 1),
        "Dari total butir": total_butir,
    })
tabel_n = pd.DataFrame(baris)
display(tabel_n)
```
